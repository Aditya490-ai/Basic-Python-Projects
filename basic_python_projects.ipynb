```python
## ------------------------------------------Expense Tracker Project---------------------------------------




print("-----------welcome in the expense tracker application--------------")

track=[]
while True:
    print("\n\n\n1. Add expenses")
    print("2. View expenses")
    print("3. Total expenses")
    print("4. Exit \n")

    choice = input("enter your choice")
    if choice == "1":
        print("--------------enter your expenses----------------") 
        date = input("enter the date")
        category=input("enter the category")
        price=int(input("enter the price"))

        expense = {
            "date":date,
            "category":category,
            "price":price,

        }
        
        track.append(expense)

        
        print("expense successfully added")

    elif choice == "2":
        if len(track)==0:
            print("No expenses added")

        else:
            print("-----view all expenses---------")

            count = 1
            for abc in track:
                print(
                f" track    {count} {abc['date']},"
                f" {abc['category']},"
                f" {abc['price']}"
                )
                count+=1
    elif choice =="3":
        print("--------total of all expenses-----------")
        total = 0
        for abc in track:
            total  += abc["price"]
            print("\n total expense is=",total)
        else:
            print("no any expenses in the expenselist")


    elif choice == "4":
        print("exit successfuly! Have a good day...")
        break
    
```
